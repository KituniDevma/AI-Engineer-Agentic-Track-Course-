```python
# The imports

import os
import requests
from dotenv import load_dotenv
from openai import AsyncOpenAI
from openai.types.responses import ResponseTextDeltaEvent
from agents import Agent, Runner, trace, function_tool, SQLiteSession, OpenAIChatCompletionsModel, set_tracing_disabled

load_dotenv(override=True)
set_tracing_disabled(True)

GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
gemini_client = AsyncOpenAI(base_url=GEMINI_BASE_URL, api_key=os.getenv("GOOGLE_API_KEY"))
gemini_model = OpenAIChatCompletionsModel(model="gemini-3.5-flash", openai_client=gemini_client)
```

## Sidenote

The actual name of this framework on the official Python index pypi.org is `openai-agents`

So for your own projects in the future, you would do:

`pip install openai-agents`  
or  
`uv add openai-agents`

followed by

`from agents import Agent, Runner, trace`

Beware that doing a `pip install agents` would install something completely different - an older reinforcement learning library.


```python

# Make an agent with name, instructions, model

agent = Agent(name="Jokester", instructions="You are a joke teller", model=gemini_model)
```

```python
# Run the joke with Runner.run(agent, prompt)

result = await Runner.run(agent, "Tell a joke about Autonomous AI Agents")
```

```python
# Here is the final output

print(result.final_output)
```

```python
# Here is the detail of the LLM calls

result.to_input_list()
```

## Adding Observability with a trace

```python
with trace("Telling a joke"):
    result = await Runner.run(agent, "Tell a joke about Autonomous AI Agents")
print(result.final_output)
```

## Now go and look at the trace

https://platform.openai.com/traces

```python
# Streaming

result = Runner.run_streamed(agent, input="Please tell me 5 jokes about AI Agents.")
async for event in result.stream_events():
    if event.type == "raw_response_event" and isinstance(event.data, ResponseTextDeltaEvent):
        print(event.data.delta, end="", flush=True)
```

## Part 2: Adding a tool

```python
pushover_user = os.getenv("PUSHOVER_USER")
pushover_token = os.getenv("PUSHOVER_TOKEN")
pushover_url = "https://api.pushover.net/1/messages.json"

if pushover_user:
    if pushover_user.startswith("u"):
        print("Pushover user found and looks good")
    else:
        print("Pushover user found but doesn't start with u")
else:
    print("Pushover user not found")

if pushover_token:
    if pushover_token.startswith("a"):
        print("Pushover token found and looks good")
    else:
        print("Pushover token found but doesn't start with a")
else:
    print("Pushover token not found")
```

```python
# Remember this?

def push(message):
    print(f"Push: {message}")
    payload = {"user": pushover_user, "token": pushover_token, "message": message}
    requests.post(pushover_url, data=payload)
```

```python
push("HEY!!")
```

```python
push
```

```python
# Now this:

@function_tool
def push_tool(message: str) -> str:
    """ Send the given message to the user as a push notification """
    payload = {"user": pushover_user, "token": pushover_token, "message": message}
    result = requests.post(pushover_url, data=payload).status_code
    return f"Push sent with API status code {result}"
```

```python
push_tool
```

```python
push_tool.description
```

```python

notifier = Agent(name="Notifier", model=gemini_model, instructions="You notify the user upon request", tools=[push_tool])
```

```python
with trace("Pizza has arrived"):
    result = await Runner.run(notifier, "Notify the user that the pizza is here")

print(result.final_output)
```

## Part 3: Sessions (memory)

Within a Runner.run() application level turn, the conversation history is maintained.

But each call to Runner.run() is a fresh start.

Let's see that:

```python
agent = Agent(name="Assistant", model=gemini_model)
```

```python
response = await Runner.run(agent, "Hi there. My name is Ed.")
print(response.final_output)
```

```python
response = await Runner.run(agent, "What's my name?")
print(response.final_output)
```

## Memory approach 1 - just manually pass in the list of dicts

```python
response = await Runner.run(agent, "Hi there. My name is Ed.")
print(response.final_output)
```

```python
response.to_input_list()
```

```python
next_input = response.to_input_list() + [{"role": "user", "content": "What's my name?"}]
next_input
```

```python
response = await Runner.run(agent, next_input)
print(response.final_output)
```

## Another approach - use OpenAI Agents SDK built in SQLLite session

```python
# This is created in-memory
# For an on-disk memory, use SQLiteSession("12345", "memory.db")

session = SQLiteSession("12346")
```

```python
response = await Runner.run(agent, "Hi there. My name is Ed.", session=session)
print(response.final_output)
```

```python
response = await Runner.run(agent, "What's my name?", session=session)
print(response.final_output)
```

## Exercise: Week 1 digital twin, with the Agents SDK

The week 1 twin needed a handwritten loop: call the model, run tools, append the results, call again.
Here that loop is `Runner.run`. The tools are ordinary functions with `@function_tool`, and the model is the Gemini model from the top of this notebook.


```python
from pathlib import Path
from pypdf import PdfReader

candidates = [
    Path("../1_foundations/twin"),
    Path("1_foundations/twin"),
    Path("twin"),
]
twin_dir = next(path for path in candidates if (path / "summary.txt").exists())

summary = (twin_dir / "summary.txt").read_text(encoding="utf-8")
reader = PdfReader(twin_dir / "linkedin.pdf")
linkedin = ""
for page in reader.pages:
    text = page.extract_text()
    if text:
        linkedin += text

twin_instructions = f"""
# Your role

You are a digital twin on a website, chatting with visitors.
You represent this person and answer questions about their career, background, skills, and experience.
If asked, say clearly that you are an AI digital twin, not the person.

# Person

{summary}

# LinkedIn

{linkedin}

# Rules

Be professional, as if talking to a potential client or future employer.
If a question is unrelated to career, background, skills, or experience, steer back to professional topics.
Do not answer questions about food, hobbies, or personal taste, even if they appear in the summary.
If you do not know the answer, call record_unknown_question, then say you do not know. Never invent facts.
If the visitor wants to get in touch and gives an email, call record_user_details, then confirm it was recorded.
Reply in markdown, with no code blocks.
"""
```

```python
@function_tool
def record_user_details(email: str, name: str = "Name not provided", notes: str = "not provided") -> str:
    """Record that a visitor wants to get in touch and shared an email address."""
    with open("twin_contacts.txt", "a", encoding="utf-8") as f:
        f.write(f"{name} | {email} | {notes}\n")
    return "Recorded"


@function_tool
def record_unknown_question(question: str) -> str:
    """Record a question the twin could not answer from the profile."""
    with open("twin_contacts.txt", "a", encoding="utf-8") as f:
        f.write(f"UNKNOWN | {question}\n")
    return "Recorded"


twin = Agent(
    name="Digital Twin",
    instructions=twin_instructions,
    model=gemini_model,
    tools=[record_user_details, record_unknown_question],
)
```

```python
session = SQLiteSession("digital-twin")

reply = await Runner.run(twin, "What kind of engineering work do you do?", session=session)
print(reply.final_output)
```

```python
reply = await Runner.run(twin, "What is your favorite pizza topping?", session=session)
print(reply.final_output)
```

```python
reply = await Runner.run(
    twin,
    "Please contact me. My name is Sam and my email is sam.visitor@example.com. I want to talk about an agent project.",
    session=session,
)
print(reply.final_output)
print("--- contacts file ---")
print(Path("twin_contacts.txt").read_text(encoding="utf-8"))
```
